## 1. Configure data sources

Each source is defined with its URL, output filename and purpose. Keeping this configuration separate from the scraping logic makes the collection pipeline easier to maintain and extend.

```python
TABLES = {
    "standard": {
        "url": "https://fbref.com/en/comps/Big5/stats/players/Big-5-European-Leagues-Stats",
        "filename": "standard.csv",
        "purpose": "Player identity, age, position, squad and general playing statistics",
    }
}
```

```python
TABLES["standard"]
```

```python
HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 "
        "(Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 "
        "(KHTML, like Gecko) "
        "Chrome/138.0 Safari/537.36"
    )
}
```

```python
STANDARD_URL = TABLES["standard"]["url"]

response = requests.get(
    STANDARD_URL,
    headers=HEADERS,
    timeout=30,
)
```

```python
print("Status code:", response.status_code)
print("Content type:", response.headers.get("Content-Type"))
print("Response length:", len(response.text))
```
